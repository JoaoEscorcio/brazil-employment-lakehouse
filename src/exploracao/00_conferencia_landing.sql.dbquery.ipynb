```sql
-- Conferência da carga histórica no volume landing.raw
-- Compare os resultados com ingestao_local/contagens.txt

-- 1) CAGED: linhas por tipo de arquivo e competência (esperado: 36 linhas)
SELECT tipo, competencia, COUNT(*) AS linhas
FROM parquet.`/Volumes/brazil_employment/landing/raw/caged/`
GROUP BY ALL
ORDER BY competencia, tipo;

-- 2) RAIS: total de vínculos (esperado: 1.931.903)
SELECT COUNT(*) AS linhas
FROM parquet.`/Volumes/brazil_employment/landing/raw/rais/`;

-- 3) PNAD: pessoas por trimestre (esperado: 6 linhas, 46.644 no total)
SELECT ano, trimestre, COUNT(*) AS linhas
FROM parquet.`/Volumes/brazil_employment/landing/raw/pnad/`
GROUP BY ALL
ORDER BY ano, trimestre;
```
