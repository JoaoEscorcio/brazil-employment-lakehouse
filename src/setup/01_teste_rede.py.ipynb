```python
import socket

fontes = {
    "CAGED e RAIS (FTP do MTE, porta 21)": ("ftp.mtps.gov.br", 21),
    "PNAD (servidor do IBGE, https)": ("ftp.ibge.gov.br", 443),
}
for nome, (host, porta) in fontes.items():
    try:
        socket.create_connection((host, porta), timeout=10).close()
        print("ALCANÇA  ", nome)
    except Exception as e:
        print("BLOQUEADO", nome, "→", type(e).__name__)
```

```python
import ftplib, urllib.request

# 1) CAGED: baixa o menor arquivo (EXC de jan/2026, ~120 KB) para a memória
dados = bytearray()
with ftplib.FTP("ftp.mtps.gov.br", timeout=60, encoding="latin-1") as ftp:
    ftp.login()
    ftp.cwd("pdet/microdados/NOVO CAGED/2026/202601")
    ftp.retrbinary("RETR CAGEDEXC202601.7z", dados.extend)
print(f"CAGED: baixou {len(dados):,} bytes")

# 2) PNAD: lê o início da página de arquivos do IBGE
url = "https://ftp.ibge.gov.br/Trabalho_e_Rendimento/Pesquisa_Nacional_por_Amostra_de_Domicilios_continua/Trimestral/Microdados/2026/"
with urllib.request.urlopen(url, timeout=60) as r:
    print(f"PNAD: status {r.status}, leu {len(r.read(2000)):,} bytes")
```
